1. What is a RESTful API?

A **RESTful API (Representational State Transfer API)** is a way for different software applications to communicate with each other over the internet using **HTTP (like websites do)**. It follows a set of architectural rules (REST principles) to make communication simple, scalable, and fast.

---

 Key Concepts of RESTful API

| Concept       | Meaning                                                    |
| ------------- | ---------------------------------------------------------- |
| **Client**    | The system sending the request (e.g., mobile app, web app) |
| **Server**    | The system providing the service/data                      |
| **Request**   | The action the client asks (GET, POST, etc.)               |
| **Response**  | The result returned by the server (data, status message)   |
| **Resources** | Data objects like users, products, orders                  |

---

 HTTP Methods Used in RESTful APIs

| Method     | Purpose              | Example           |
| ---------- | -------------------- | ----------------- |
| **GET**    | Fetch data           | Get list of users |
| **POST**   | Create new data      | Add a new order   |
| **PUT**    | Update existing data | Edit user profile |
| **DELETE** | Remove data          | Delete a product  |

---

→ Creates a new product.

---

 Data Format

RESTful APIs usually send and receive data in:

* **JSON** (most common)
* **XML** (older format)

**Example JSON Response:**

```json
{
  "id": 1,
  "name": "Laptop",
  "price": 55000
}

2. What is Flask, and why is it popular for building APIs?

**Flask** is a **lightweight and flexible web framework for Python** used to build web applications and **RESTful APIs**. It provides the essential tools to build a server but leaves most decisions (design, architecture, libraries) to the developer — which makes it simple but highly customizable.

---

 Why Flask is Popular for Building APIs

| Reason                           | Explanation                                                             |
| -------------------------------- | ----------------------------------------------------------------------- |
| **Lightweight & simple**         | It doesn't force structure—easy to get started and learn.               |
| **Fast development**             | You can build APIs quickly with minimal code.                           |
| **Flexible**                     | Developers can choose their own components (database, structure, etc.). |
| **Built-in development server**  | Helps in easy testing during development.                               |
| **Supports RESTful design**      | Very easy to create REST API endpoints.                                 |
| **Large community & extensions** | Many plugins available (Flask-SQLAlchemy, Flask-RESTful, JWT, etc.).    |

---

 Example: A simple API using Flask

```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/hello', methods=['GET'])
def hello():
    return jsonify({"message": "Hello, Flask API!"})

if __name__ == "__main__":
    app.run(debug=True)
```

When run, visiting:

```
http://localhost:5000/hello
```

returns:

```json
{
  "message": "Hello, Flask API!"
}
```

4. What is the purpose of the @app.route() decorator in Flask?

In Flask, the **@app.route() decorator** is used to **connect a URL (route) with a specific function** in the application.

 What it does

When someone visits a URL in the browser (e.g., `/home`, `/users`, `/login`), Flask needs to know **which function should run** to generate the response.
The `@app.route()` decorator performs that mapping.

---

 Example

```python
@app.route('/hello')
def greeting():
    return "Hello, World!"
```

This means:

* When the user goes to `http://localhost:5000/hello`
* Flask will execute the `greeting()` function and return its output.

5. What is the role of Flask-SQLAlchemy?

**Flask-SQLAlchemy** is an extension for Flask that makes it easy to work with **databases using SQLAlchemy**.

 Role / Purpose

Flask-SQLAlchemy provides:

* A high-level interface to connect Flask apps with a database
* Tools to define tables and models using Python classes (instead of writing raw SQL)
* Easy methods to insert, update, delete, and query data

---

 Why it is used

| Benefit            | Description                                                         |

| Easy integration   | Connects Flask with different DBs (SQLite, MySQL, PostgreSQL, etc.) |
| Clean and scalable | Helps maintain clean and structured database code                   |

To return **JSON responses in Flask**, you typically use the **`jsonify()`** function, which converts Python data into JSON and sets the correct content type.

---

 Example

```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/user')
def user():
    data = {"id": 1, "name": "Aarju", "role": "student"}
    return jsonify(data)

if __name__ == "__main__":
    app.run(debug=True)
```

When visiting `/user`, the response will be in **JSON format**:

```json
{
  "id": 1,
  "name": "Aarju",
  "role": "student"
}
```

---

 You can also return status codes

```python
return jsonify({"message": "Created successfully"}), 201
```

---

 Or include headers if needed

```python
return jsonify({"error": "Unauthorized"}), 401
```

8. How do you handle POST requests in Flask?

To handle **POST requests in Flask**, you define a route that allows the **POST** method and then access the data sent in the request.

---

 Basic Example (receiving JSON data)

```python
from flask import Flask, request, jsonify

app = Flask(__name__)

@app.route('/add', methods=['POST'])
def add_data():
    data = request.get_json()   # Get JSON payload from the POST request
    return jsonify({
        "message": "Data received",
        "data": data
    }), 201

if __name__ == "__main__":
    app.run(debug=True)
```

When a POST request sends:

```json
{
  "name": "Aarju",
  "age": 21
}
```

The API returns:

```json
{
  "message": "Data received",
  "data": {
    "name": "Aarju",
    "age": 21
  }
}
```

---

 Handling form-data (from HTML forms)

```python
@app.route('/login', methods=['POST'])
def login():
    username = request.form['username']
    password = request.form['password']
    return f"Logged in as {username}"
```

---

 Key Functions Used

| Function             | Purpose                  |
| -------------------- | ------------------------ |
| `methods=['POST']`   | Allows POST on the route |
| `request.get_json()` | Fetch JSON body          |
| `request.form`       | Fetch form data          |
| `request.files`      | Fetch uploaded files     |

9. How do you handle errors in Flask (e.g., 404)?

In Flask, you handle errors like **404 (Not Found)** or **500 (Server Error)** by using **error handler functions** with the `@app.errorhandler()` decorator.

---

 Example: Handling 404 Error

```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.errorhandler(404)
def not_found(error):
    return jsonify({"error": "Resource not found"}), 404
```

If a user visits an undefined route, this response will be shown in JSON.

---

 Handling other errors (example: 500)

```python
@app.errorhandler(500)
def internal_error(error):
    return jsonify({"error": "Internal server error"}), 500
```

---

 Handling custom error response

You can raise an error manually:

```python
from werkzeug.exceptions import BadRequest

@app.route('/check')
def check():
    raise BadRequest("Invalid request data")
```

10. How do you structure a Flask app using Blueprints?

Structuring a Flask app using **Blueprints** helps divide the application into multiple reusable modules, making it organized and scalable — especially for larger projects or APIs.

---

 What is a Blueprint?

A **Blueprint** in Flask is a way to group routes, views, and functions into separate components so they can be registered later in the main application.

---

 Typical Folder Structure

```
myapp/
│
├── app.py
├── requirements.txt
│
├── users/
│   ├── __init__.py
│   └── routes.py
│
└── products/
    ├── __init__.py
    └── routes.py
```

---

 Step 1: Create a Blueprint (example: users)

**users/routes.py**

```python
from flask import Blueprint, jsonify

users_bp = Blueprint('users', __name__)

@users_bp.route('/users')
def get_users():
    return jsonify({"message": "Users endpoint"})
```

---

 Step 2: Register the Blueprint in the main app

**app.py**

```python
from flask import Flask
from users.routes import users_bp
from products.routes import products_bp

app = Flask(__name__)

# Register blueprints
app.register_blueprint(users_bp)
app.register_blueprint(products_bp)

if __name__ == "__main__":
    app.run(debug=True)
```

---

Another Blueprint Example

**products/routes.py**

```python
from flask import Blueprint, jsonify

products_bp = Blueprint('products', __name__)

@products_bp.route('/products')
def get_products():
    return jsonify({"message": "Products endpoint"})
```

---

 Result

Routes become automatically available:

```
/users
/products
```

---

 Benefits of Using Blueprints

| Advantage           | Description                                             |
| ------------------- | ------------------------------------------------------- |
| Better organization | Separates feature modules (users, products, auth, etc.) |
| Scalability         | Easy to add new modules without messing up the app      |
| Reusability         | Blueprints can be reused across multiple apps           |
| Cleaner code        | Prevents one huge `app.py` file                         |
